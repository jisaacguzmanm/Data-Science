
# TF-IDF (Term Frequency - Inverse Document Frequency)
Esta es una tecnica utilizada en el procesamiento de lenguaje natural (NLP) y minería de texto. Su objetivo principal es representar un conjunto de documentos de manera que las palabras más relevantes (es decir, las que mejor caracterizan un documento o un conjunto de documentos) tengan mayor peso, mientras que las palabras comunes (que se repiten mucho en todos los documentos, como "de", "y", "la") tengan menos peso.

A continuación, explicamos en detalle qué es TF-IDF, cómo se calcula y cómo se aplica:
Componentes de TF-IDF:

IDF (Inverse Document Frequency - Frecuencia Inversa de Documentos):

Mide la importancia de un término en un conjunto de documentos (corpus).

Si un término aparece en muchos documentos, su valor IDF será bajo, ya que no es informativo. Si un término aparece en pocos documentos, su valor IDF será alto, ya que es más distintivo.

Se calcula usando la fórmula:

${IDF}(t) =log( \frac{\text{Número total de documentos}}{\text{Número de documentos que contienen el término } t})$

Ejemplo: Si tienes un corpus con 100 documentos y la palabra "ciencia" aparece en 10 de esos documentos, el IDF de "ciencia" será:

$$IDF("ciencia")=log⁡(\frac{100}{10})=log⁡(10)≈1$$


TF-IDF (Term Frequency-Inverse Document Frequency):

El valor de TF-IDF de un término es simplemente el producto de TF y IDF.

Esta métrica le da más peso a las palabras que son frecuentes en un documento pero poco comunes en el resto del corpus.

Fórmula:
        $$TF-IDF(t,d)=TF(t,d)×IDF(t)$$

Ejemplo: Si la palabra "ciencia" tiene una TF de 0.05 y un IDF de 1, el TF-IDF sería:
        $$0.05×1=0.05$$


```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Conjunto de documentos de ejemplo
corpus = [
    "El procesamiento de lenguaje natural es fascinante",
    "Las redes neuronales son importantes en PLN",
    "Aprender sobre PLN es crucial en la Maestría de Ciencia de Datos"
]
```

Se aplica la técnica de TF-IDF usando la clase TfidfVectorizer de la librería scikit-learn, que facilita la conversión de un conjunto de textos en una matriz numérica basada en los valores de TF-IDF de cada término.

**TfidfVectorizer()**: Crea una instancia del vectorizador TF-IDF. Esta clase toma un conjunto de documentos (corpus) y convierte las palabras en una representación numérica basada en la técnica TF-IDF. Este vectorizador va a aprender de los documentos y calcular los valores de TF e IDF para cada palabra.

fit_transform(corpus): Aquí, se aplica el proceso de ajuste y transformación al conjunto de textos.

  *fit*: Se ajusta el modelo a los datos, es decir, calcula la matriz IDF para todos los términos en el corpus.

  *transform:* Convierte el texto del corpus en una matriz dispersa (es decir, una matriz con muchos valores en cero) donde cada fila es un documento y cada columna es un término, con los valores correspondientes a los valores de TF-IDF.

```python

# Aplicar TF-IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

# Mostrar resultados
print("Vocabulario:", vectorizer.get_feature_names_out())
```

```python
print("Matriz TF-IDF:")
print(X.toarray())
```


#Word embeddings para NLP  (usando Gensim)

¿Qué son los word embeddings?

Word embeddings son representaciones vectoriales de palabras en un espacio multidimensional, donde las palabras con significados similares están representadas por vectores cercanos entre sí. A diferencia de los métodos tradicionales de representación de texto como el "bag-of-words", los embeddings capturan relaciones semánticas entre las palabras.

Por ejemplo, "rey" y "reina" pueden tener vectores que están muy cerca uno del otro, mientras que "rey" y "perro" estarán más distantes en el espacio vectorial.

Los word embeddings permiten representar palabras en un formato que es fácil de manejar para las máquinas, ya que transforman el texto en números, manteniendo las relaciones semánticas entre las palabras.

Existen varios algoritmos para entrenar embeddings, como Word2Vec, GloVe y FastText. Estos algoritmos aprenden los embeddings a partir de grandes volúmenes de texto, basándose en el contexto en el que las palabras aparecen.


# Gensim

Gensim es una librería de Python especializada en el modelado de temas y la representación de texto, particularmente en la implementación de algoritmos de word embeddings como Word2Vec.

Gensim está optimizada para trabajar con grandes corpus de texto, ya que se basa en un modelo de procesamiento en memoria eficiente. Además, es muy útil para tareas como:

a. Modelado de temas (como LDA o Latent Dirichlet Allocation)
b. Word embeddings (Word2Vec, FastText, etc.).
c. Transformación de texto en representaciones vectoriales (TF-IDF, LSI, etc.).

Gensim también facilita la carga de modelos preentrenados, lo cual es útil cuando se trabaja con embeddings que ya han sido entrenados con grandes corpus de texto, como en este caso, donde estamos utilizando un modelo de embeddings en español.


Embendding y el procesamiento de lenguaje natural (NLP)

En el contexto de NLP (Natural Language Processing), los word embeddings permiten a las máquinas comprender el significado de las palabras no solo por su forma, sino también por su contexto y significado. Usar embeddings ayuda a mejorar muchas tareas en NLP, como:

Clasificación de texto: Usar embeddings como características


para entrenar modelos de clasificación.

  * Análisis de sentimientos: Comparar la similitud entre las palabras para identificar la polaridad de un texto.
  * Traducción automática: Los embeddings permiten que los modelos aprendan relaciones semánticas entre palabras en diferentes idiomas.

Este código descarga y carga un archivo de vectores de palabras preentrenados (en formato .vec) que ha sido entrenado con grandes corpus de texto en español. Estos vectores ya contienen las representaciones de las palabras, y el código de abajo permite cargarlas y verificarlas.

La siguiente celda de código verifica si el archivo de vectores preentrenados ya está descargado en tu entorno. Si no está presente, lo descarga desde una URL proporcionada. El archivo es un modelo de embeddings en español.

KeyedVectors.load_word2vec_format(): Esta función de Gensim se usa para cargar modelos de vectores preentrenados. En este caso, el archivo .vec que descargamos es un modelo en formato texto, por lo que binary=False indica que el archivo no está en formato binario (si estuviera en binario, el parámetro sería True).
wordvectors.key_to_index: Este atributo contiene todos los términos que están presentes en el modelo y nos da acceso al número total de vectores cargados.

```python
from gensim.models import KeyedVectors

# Ruta al archivo de vectores preentrenados (en formato de texto)
wordvectors_file = 'embeddings-m-model.vec'

# Cargar el archivo de vectores con KeyedVectors en formato de texto
wordvectors = KeyedVectors.load_word2vec_format(wordvectors_file, binary=False)

# Verificar que se haya cargado correctamente
print(f"Se cargaron {len(wordvectors.key_to_index)} vectores.")
```

# Visualización 2D de Vectores de Palabras

Una forma muy útil de entender las relaciones entre palabras es visualizarlas en un espacio 2D. Para esto, reducimos las 300 dimensiones del vector de cada palabra a solo 2 dimensiones, utilizando técnicas como PCA o t-SNE.

Para realizar una visualización 2D de un conjunto de vectores de palabras seleccionadas de un modelo de Word2Vec (en formato .vec), debemos:

1. Seleccionamos un conjunto de palabras agrupadas en categorías como animales, colores, profesiones, etc.

2. Obtenemos los vectores para estas palabras del modelo de Word2Vec.

3. Reducimos la dimensionalidad de los vectores a 2D utilizando PCA o t-SNE.

4. Visualizamos los vectores en un gráfico de dispersión, donde cada grupo de palabras está representado por un color y marcador diferente.

##Paso 1: Cargar el modelo y seleccionar las palabras

Primero, ya tienes el código para cargar el modelo de Word2Vec. Luego, seleccionamos un conjunto de 300 palabras que pertenecen a diferentes grupos semánticos (por ejemplo, animales, colores, países, etc.). Es importante que las palabras seleccionadas sean de distintos grupos para que la visualización 2D sea interesante y clara.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import itertools
```

```python
# Selección de palabras, agrupadas en grupos semánticamente similares
groups = {
    'animales': ['perro', 'gato', 'elefante', 'jirafa', 'loro'],
    'transporte': ['bus', 'avión', 'carro', 'tren', 'camión'],
    'frutas': ['manzana', 'plátano', 'naranja', 'uva', 'mango'],
    'países': ['españa', 'colombia', 'italia', 'alemania', 'portugal'],
    'profesiones': ['alumno', 'ingeniero', 'doctor', 'arquitecto', 'profesor']
}
```

##Paso 2: Obtener los vectores de las palabras seleccionadas

Una vez cargado el modelo y seleccionadas las palabras, obtenemos los vectores correspondientes.

##Paso 3: Reducir la dimensionalidad

Usamos t-SNE (t-Distributed Stochastic Neighbor Embedding) o PCA (Principal Component Analysis) para reducir la dimensionalidad de los vectores de 300 dimensiones a 2 dimensiones, lo que nos permitirá visualizarlos en un gráfico 2D.

```python
# Obtener los vectores para las palabras seleccionadas
words = list(itertools.chain(*groups.values()))
vectors = [wordvectors[word] for word in words if word in wordvectors]

# Reducir la dimensionalidad a 2D usando PCA
pca = PCA(n_components=2)
result_pca = pca.fit_transform(vectors)
```

```python


# Crear el gráfico
plt.figure(figsize=(10, 8))

# Colores y marcadores para cada grupo
colors = ['red', 'blue', 'green', 'purple', 'orange']
markers = ['o', 's', '^', 'D', 'v']

# Graficar los puntos para cada grupo
for i, (group_name, group_words) in enumerate(groups.items()):
    indices = [words.index(word) for word in group_words]
    plt.scatter(result_pca[indices, 0], result_pca[indices, 1],
                color=colors[i], label=group_name, marker=markers[i], s=100)

    # Etiquetas de las palabras
    for j, word in enumerate(group_words):
        plt.text(result_pca[indices[j], 0] + 0.01, result_pca[indices[j], 1] + 0.01, word, fontsize=12)

# Añadir leyenda y mostrar el gráfico
plt.title('Visualización 2D de Vectores de Palabras', fontsize=16)
plt.legend(loc='best')
plt.grid(True)
plt.show()
```

Proceso

Descargar el archivo del modelo FastText: El código proporciona la URL para descargar el modelo FastText preentrenado para el idioma español. El modelo tiene 300 dimensiones para cada palabra y es parte del corpus de "Common Crawl".

```python
import urllib.request
import os

# URL del archivo (el archivo contiene los vectores de palabras preentrenados)
url = 'https://dl.fbaipublicfiles.com/fasttext/vectors-crawl/cc.es.300.bin.gz'
# Nombre del archivo donde se almacenará el modelo de vectores
downloaded_file = 'cc.es.300.bin.gz'
```

```python

# Comprobar si el archivo ya existe antes de descargarlo
if not os.path.exists(downloaded_file):
    print("Descargando el archivo...")
    # Save the downloaded file with the correct filename
    urllib.request.urlretrieve(url, downloaded_file)
    print("Descarga completa.")
else:
    print(f"El archivo {downloaded_file} ya existe. Omitiendo descarga.")
```

Descomprimir el archivo: El archivo descargado está comprimido en formato .gz. El código menciona que después de la descarga, es necesario descomprimirlo. Esta descompresión se puede hacer usando la librería gzip de Python o a través de otros métodos en el sistema operativo.

Vamos descomprimirlo con Python:

```python
import gzip
import shutil

# Descomprimir el archivo
with gzip.open(downloaded_file, 'rb') as f_in:
    with open(downloaded_file[:-3], 'wb') as f_out:
        shutil.copyfileobj(f_in, f_out)
```

```python
from scipy.spatial.distance import cosine

# Lista de palabras
words = ['paz', 'igualdad', 'inclusión', 'petro']

# Obtener los vectores de todas las palabras
word_vectors = [ft.get_word_vector(word) for word in words]

# Calcular la similitud coseno entre todas las combinaciones de palabras
similarities = []

# Comparamos cada palabra con las demás
for i, word_vec_1 in enumerate(word_vectors):
    similarity_sum = 0
    count = 0
    for j, word_vec_2 in enumerate(word_vectors):
        if i != j:
            similarity = 1 - cosine(word_vec_1, word_vec_2)  # Similaridad coseno
            similarity_sum += similarity
            count += 1
    avg_similarity = similarity_sum / count
    similarities.append((words[i], avg_similarity))

# Ordenar las palabras por similitud promedio (la de menor similitud es la que "no encaja")
similarities.sort(key=lambda x: x[1])

# Mostrar la palabra que menos "encaja"
print(f"La palabra que no encaja es: {similarities[0][0]}")
```

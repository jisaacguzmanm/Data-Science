## Roles en la Interacción con LLMs (desde Python)

Cuando se interactúa con un Large Language Model (LLM) a través de un script de Python, generalmente se manejan varios 'roles' para estructurar la conversación y guiar el comportamiento del modelo. Los roles más comunes son:

1.  **Rol de Usuario (`user`)**:
    *   Representa la entrada o la pregunta directa que el usuario final le hace al modelo. Es el 'prompt' principal. Lo que se espera que el modelo responda.

2.  **Rol de Sistema (`system`)**:
    *   Proporciona instrucciones o contexto inicial al modelo para establecer su comportamiento, personalidad, o formato de salida deseado. El modelo intenta seguir estas directrices a lo largo de la conversación. Por ejemplo: "Eres un asistente amigable y útil que responde de forma concisa."

3.  **Rol de Asistente (`assistant`)**:
    *   Representa las respuestas generadas por el propio LLM. Es la salida del modelo en respuesta a los mensajes del usuario (y siguiendo las directrices del sistema).

Estos tres roles (Usuario, Sistema, Asistente) son fundamentales en la mayoría de las APIs de LLMs (como Gemini, OpenAI, etc.) para construir un diálogo estructurado y mantener el contexto de la conversación. Algunos modelos más avanzados también pueden introducir roles adicionales, como 'tool' o 'function', cuando se integran capacidades de llamada a herramientas o funciones externas.

3.  **Rol de Asistente (La Respuesta del LLM):**
    *   Este rol es la respuesta generada por el propio LLM. Es el resultado de procesar la entrada del `user` bajo las directrices (explícitas o implícitas) del `system`.
    *   **Ejemplo para este caso:** El asistente generará un resumen conciso y bien estructurado de toda la información proporcionada (página principal + subpáginas), basándose en las instrucciones iniciales del sistema y la solicitud del usuario.

# Ejemplo de Script OpenAI con roles (system y user)

```python
from openai import OpenAI

# 1. Crear cliente con tu API Key
client = OpenAI(api_key="TU_API_KEY")

# 2. Definir mensajes con roles
response = client.chat.completions.create(
    model="gpt-4o-mini",  # modelo económico
    messages=[
        {
            "role": "system",
            "content": "Eres un experto en gastronomía colombiana. Explica de forma clara y sencilla."
        },
        {
            "role": "user",
            "content": "¿Cómo se hace el mute santandereano?"
        }
    ],
    temperature=0.5
)

# 3. Mostrar respuesta
print("\n--- RESPUESTA ---\n")
print(response.choices[0].message.content)
```

## Configuración de Gemini con la API Key

Para interactuar con el modelo Gemini, primero necesitas configurar la librería de Google Generative AI con tu API Key.

1.  **Importar librerías necesarias:** `google.generativeai` para la API de Gemini y `google.colab.userdata` para acceder a tu secreto guardado.
2.  **Obtener la API Key:** Accede a la `GOOGLE_API_KEY` que guardaste en los secretos de Colab.
3.  **Configurar la API:** Inicializa la librería `genai` con tu API Key.
4.  **Inicializar el modelo:** Crea una instancia del modelo Gemini que deseas usar (por ejemplo, `'gemini-1.5-flash'`).

````python
import google.generativeai as genai
from google.colab import userdata

# Obtener la API Key de los secretos de Colab
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')

# Configurar la API con tu clave
genai.configure(api_key=GOOGLE_API_KEY)

# Inicializar el modelo Gemini que deseas usar
gemini_model = genai.GenerativeModel("gemini-2.5-flash") # Cambiado de 'gemini-pro' a 'gemini-pro-latest'

print("Modelo Gemini configurado y listo para usar.")
````

## Hacer una pregunta al LLM y obtener una respuesta

Una vez que el modelo está configurado, puedes hacerle preguntas directamente. Utilizaremos el método `generate_content` del modelo y luego imprimiremos la propiedad `text` de la respuesta para ver la salida.

````python
# Pregunta al LLM
question = "¿cual la cantidad de hipotatamos que hay en colombia?"
print(f"Pregunta: {question}")

# Obtener la respuesta del LLM
response = gemini_model.generate_content(question)

# Imprimir la respuesta
print(f"Respuesta del LLM: {response.text}")
````

# Diferencia entre Gemini y ChatGPT (roles)

## ChatGPT (OpenAI)
Utiliza roles definidos:
- `system`: establece el comportamiento del modelo (por ejemplo, "Eres un profesor")
- `user`: contiene la pregunta o instrucción
- `assistant`: corresponde a la respuesta del modelo

Se basa en una estructura de conversación formal con separación clara de funciones.

---

## Gemini (Google)
No utiliza el rol `system`.

Maneja principalmente:
- `user`: incluye tanto las instrucciones como la tarea
- `model`: representa la respuesta generada (de forma interna)

En este caso, toda la información se envía como un único prompt.

---

## Idea clave

- ChatGPT separa el comportamiento del modelo y la tarea a ejecutar  
- Gemini integra ambos elementos en un solo mensaje

````python
system_instruction = "Eres un chef de cocina colombiana de cinco estrellas, experto en la gastronomía de Santander. Tus explicaciones son concisas, elegantes y apetitosas."
user_question = "¿Cómo se prepara un mute santandereano? Dame una receta muy corta y resumida. quiero que me des en maximo 70 tokens la respuesta, es obligaotrio esta restriccion"

# Estructura del mensaje con roles
# Para una interacción más estructurada con Gemini, los mensajes se pueden pasar como una lista de diccionarios,
# donde cada diccionario representa un rol ('system', 'user', 'model' o 'assistant') y su contenido ('parts').
# Dado que el modelo actual no soporta el rol 'system' explícitamente, combinamos la instrucción del sistema
# con la pregunta del usuario en un solo rol 'user'.

prompt_combinado = system_instruction + "\n\n" + user_question
print(f"Como se ve el promt combinado: {prompt_combinado}")

prompt_parts = [
  {"role": "user", "parts": [prompt_combinado]}
]

print(f"Pregunta al Chef: {user_question}")

# Obtener la respuesta del LLM con el modelo ya configurado
response_chef = gemini_model.generate_content(prompt_parts)

# Imprimir la respuesta del asistente (el chef)
# El rol de asistente ('model') es la respuesta generada por el LLM.
print(f"Respuesta del Chef (Asistente):\n{response_chef.text}")
````

````python
import requests
from bs4 import BeautifulSoup
import json

# 1. Pedir URL
url = input("Ingrese la URL: ")

try:
    # 2. Hacer la petición
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    # 3. Parsear HTML
    soup = BeautifulSoup(response.text, 'html.parser')

    # 4. Eliminar scripts y estilos
    for script in soup(["script", "style"]):
        script.decompose()

    # 5. Extraer texto limpio
    text = soup.get_text()

    # Limpiar líneas
    lineas = text.splitlines()
    lineas_limpias = [linea.strip() for linea in lineas if linea.strip()]

    # Unir texto final
    texto_final = "\n".join(lineas_limpias)

    # 6. Validar contenido
    if len(texto_final) == 0:
        print("No se pudo extraer texto de la página.")
    else:
        print("\n--- TEXTO EXTRAÍDO ---\n")
        print(texto_final[:1000])  # muestra solo los primeros 1000 caracteres

        print("\n--- RESUMEN ---")
        datos = {
            "longitud_total": len(texto_final),
            "preview": texto_final[:300]
        }

        print(json.dumps(datos, indent=2))

except requests.exceptions.RequestException as e:
    print("Error al acceder a la página:", e)
except Exception as e:
    print("Error inesperado:", e)
````

````python
# BLOQUE INDEPENDIENTE: uso de roles en Gemini

# 1. Definir el rol del sistema (qué tipo de asistente queremos)
rol_sistema = """
Eres un asistente experto en resumir contenido web.
Debes responder de forma clara, breve y precisa.
"""

# 2. Definir el rol del usuario (lo que el usuario pide)
rol_usuario = f"""
Resume el siguiente texto en máximo 7 oraciones:

{texto_final[:3000]}
"""

# 3. Combinar el rol del sistema y el rol del usuario en un solo prompt para el rol 'user'
prompt_combinado = rol_sistema + "\n\n" + rol_usuario

# 4. Construir el prompt estructurado con roles
prompt = [
    {"role": "user", "parts": [prompt_combinado]}
]

# 5. Enviar al modelo
respuesta = gemini_model.generate_content(prompt)

# 6. Mostrar resultado
print("\n--- RESPUESTA DEL MODELO ---\n")
print(respuesta.text)
````

````python
translation_system_instruction = "Eres un traductor experto. Traduce el texto proporcionado al inglés de manera fluida y precisa."
# Usamos la respuesta generada previamente
translation_user_question = f"Por favor, traduce el siguiente resumen al inglés:\n\n{respuesta.text}"

translation_combined_prompt = translation_system_instruction + "\n\n" + translation_user_question

translation_prompt_parts = [
    {"role": "user", "parts": [translation_combined_prompt]}
]

# Enviar al modelo para la traducción
english_summary = gemini_model.generate_content(translation_prompt_parts)

print("\n--- TRADUCCIÓN AL INGLÉS ---\n")
print(english_summary.text)
````

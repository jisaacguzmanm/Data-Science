# **9.1 Detección de objetos**

La detección de objetos es una tarea de visión por computador que no solo identifica qué objetos hay en una imagen, sino también dónde están ubicados. Combina dos elementos:

**Clasificación** (¿qué es?),

**Localización** (¿dónde está?).

*Ejemplo: *

![imangen](https://github.com/adiacla/bigdata/blob/master/deteccion.jpg?raw=true)

***Nota.*** mihaileric, RCNN para visión por computador. 2025, https://www.mihaileric.com/static/dog_with_ball-db327d9f059c1de766b4eaa6acb2039b-f3e7c.png

Red de detección de objetos aplicada a una foto puede identificar un perro, una pelota, y dibujar cuadros delimitadores (bounding boxes) alrededor de cada uno.

La detección de objetos es una tarea de visión por computador que combina:

**Regresión:** para predecir la posición y tamaño de cada objeto en la imagen, generalmente en forma de (x, y, ancho, alto).

**Clasificación:** para identificar qué clase representa cada objeto detectado (por ejemplo: persona, perro, auto).

Cada modelo de detección realiza ambas tareas simultáneamente.

Diferencia entre detección de objetos y reconocimiento de imágenes
Reconocimiento de imágenes	Detección de objetos


* **Reconocimiento de Imágenes:** Clasifica toda la imagen. Una sola etiqueta por imagen: "Esto es un perro"

* **Detección de objetos:** Detecta y localiza múltiples objetos. Múltiples objetos y clases por imagen: "Aquí hay un proo y aquí un balón"

# **Etiquetar**

Etiquetar o anotar es el proceso de dibujar un cuadro o contorno alrededor del objeto.

Asociarlo con una etiqueta de clase (ej: perro, balón). Esto permite al modelo aprender qué aspecto tiene cada objeto y cómo diferenciarlo.

Un modelo ONNX (Open Neural Network Exchange) es un archivo que representa una red neuronal entrenada en un formato interoperable y estándar, diseñado para ser usado en diferentes frameworks de deep learning.

¿Qué significa esto?
Los principales frameworks de IA como PyTorch, TensorFlow, Keras, MXNet, etc., usan sus propios formatos internos para definir y guardar modelos. ONNX fue creado para permitir que un modelo entrenado en uno de esos frameworks pueda ser:

  Exportado fácilmente a otro framework.

  Ejecutado en motores optimizados (como ONNX Runtime, TensorRT, OpenVINO, etc.).

  Usado en producción en dispositivos móviles, web, o edge.

Ventajas del formato ONNX:
Interoperabilidad:
Entrenas en PyTorch, exportas a ONNX, y puedes ejecutarlo en un entorno de producción en C++, Python o incluso en IoT.

Velocidad y eficiencia:
Herramientas como ONNX Runtime permiten una inferencia más rápida y optimizada, incluso para dispositivos con pocos recursos.

Compatibilidad con hardware acelerado:
ONNX permite usar aceleradores como GPUs, TPUs, FPGAs, etc., mediante integraciones como TensorRT, OpenVINO, etc.
